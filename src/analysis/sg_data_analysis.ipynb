```python
import pandas as pd
```

```python
#!pip install plotly

import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv('Me_SO4_all_sg.csv')
```

```python
df.head
```

```python

fig = go.Figure()
dic_sg = {1:'triclinic P1',2:'triclinic P-1', 3:'monoclinic P2', 4:'monoclinic Cc',
          5:'monoclinic P2\u2081/c', 6:'orthorhombic P2\u20812\u20812\u2081',
          7:'orthorhombic Pna2\u2081'}


# Step through every pair of columns in the DataFrame
for i in range(2, df.shape[1], 3):  # Step through every pair of columns
    col1 = df.iloc[:, i]
    col2 = df.iloc[:, i + 1]
    col3 = df.iloc[:, i - 1]
    
    # Add a scatter trace for each pair of columns
    
    fig.add_trace(go.Scatter(
        x=col2, y=col1, mode='markers',marker=dict(size=8, opacity = 0.8), name=dic_sg[(i+1)//3], customdata=col3,  # Attach multiple columns
        hovertemplate="Energy: %{y}<br>Density: %{x}<br>Num_config: %{customdata}"))

# Customize the plot
fig.update_layout(
    title='Energy vs density diagram for the Me_SO4 cage crystals',
    xaxis_title="Density / g cm\u207B\u00B3",
    yaxis_title="Relative total energy / (kJ mol\u207B\u00B9 of cage)",
    xaxis_range=[0, 1],  # Set range for x-axis
    yaxis_range=[190, 300],
    legend_title="Space group",
    legend=dict(title="Space group", x=0.85, y=0.95),
    showlegend=True,
    height=600,  # Adjust the size of the figure
    width=1000,
    template='plotly_white',  # Clean background
    margin=dict(t=50, b=50, l=50, r=50)  # Adjust margins to fit titles# Adjust size of markers
)

# Show the plot
fig.show()

# Save the plot as an image (requires Kaleido)
#fig.write_image("combined.png")



```
