```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# so3_nme4
df = pd.read_csv('xtb_SO3_NMe4_coh_energy.csv')
```

```python
column_data = df['Cohesion energy/ kj/mol']
filtered_data = column_data[(column_data > -198) & (column_data < -20)]
```

```python
plt.figure(figsize=(8, 6))  # Optional: Adjust the figure size
filtered_data.hist(bins=20, density=True, color='green', edgecolor='black')
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y * 10000:.0f}'))

# Step 4: Add labels and title
plt.xlabel('Cohesion energy/ (kJ/mol)', size=14)
plt.ylabel('Normalised counts', size=14)
#plt.title('Histogram of Your Column')
# don't show gridlines
plt.grid(False)
plt.suptitle('Histogram showing the relative distribution of cohesion energies \n of the sampled discrete SO3_NMe4 cages', y =-0.005 )
# Step 5: Show the plot
plt.show()
```

```python
# concatenate some csv files that look like so3_na_{i}.csv
dfs = []
files = [f'so3_na_{i}.csv' for i in [1, 2, 3, 9, 14, 19, 33]]
for file in files:
    df = pd.read_csv(file).iloc[:, 1:]
    df['Energies'] = ((df['Energies'])*96.5)/4
    df = df[(df['Energies'] < 10000) & (df['Energies'] > 9000) & (df['Densities'] < 1.5)]
    dfs.append(df)
combined_df = pd.concat(dfs, axis =1)
combined_df.to_csv('combined.csv', index=False)
# show df
print(combined_df)
plt.figure(figsize=(10, 6))
for i in range(0, combined_df.shape[1], 2):  # Step through every pair of columns
    col1 = combined_df.iloc[:, i]
    col2 = combined_df.iloc[:, i + 1]
    
    plt.scatter(col2, col1)  # Add legend entry for the set

# Customize the plot
plt.xlabel("\n Density / g cm\u207B\u00B3 ", size = 14)
plt.ylabel("Relative total energy / (kJ mol\u207B\u00B9 of cage)", size = 14)
#plt.title("Plot of all sets of two columns")
plt.legend(['triclinic P1','triclinic P-1', 'monoclinic P2', 'monoclinic Cc', 'monoclinic P2\u2081/c', 'orthorhombic P2\u20812\u20812\u2081', 'orthorhombic Pna2\u2081'], title = 'Space group')
#title of legend
plt.suptitle('Energy vs density diagram for the Cage 2 crystals', y =-0.05 )

plt.grid(False)
plt.savefig('combined.png')
# Show the plot
plt.show()
```

```python
df = pd.read_csv('../OH_PF6/results_xtb/xtb_energies_OH_pf6.csv')
```

```python
def plot_distribution(df: pd.DataFrame) -> None:
# Create the histogram data
    data = df['Cohesion energy/ kJ/mol'][(df['Cohesion energy/ kJ/mol'] > -250) & (df['Cohesion energy/ kJ/mol'] < 0)]
    counts, bins, _ = plt.hist(data, bins=20, density=False, color='green', edgecolor='black', alpha=0.7)

# Scale the counts so that the highest bar is at 100
    scaled_counts = (counts / max(counts)) * 100

# Clear the previous plot
    plt.clf()


# Plot the scaled histogram
    plt.bar(bins[:-1], scaled_counts, width=(bins[1] - bins[0]), color='green', edgecolor='black', align='edge', alpha=0.7)

# Set the y-axis formatter and labels
    plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0f}'))
    plt.xlabel('Cohesion energy/ (kJ/mol)', size=14)
    plt.ylabel('Normalised counts', size=14)

# Remove gridlines and add a title
    plt.grid(False)
    plt.suptitle('Histogram showing the relative distribution of cohesive energies \n of the sampled discrete OH_SO4 cages', y=-0.005)

# Show the plot
    plt.show()
```

```python
plot_distribution(df)
```
