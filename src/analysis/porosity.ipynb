```python
import pandas as pd
import plotly.express as px
```

```python
def plot_1_cage_per_cell_Di(csv_path):
    df = pd.read_csv(csv_path)

    fig = px.scatter(
        df,
        x="density",
        y="energy",
        color="Di",
        color_continuous_scale="viridis",
        hover_data=["idx", "PLD"],
        title=f'{csv_path}_Di',
        labels={
            "density": "Density/ g cm\u207b\u00b3",
            "energy": "Energy per cage/ eV",
            "Di": "Di (Å)"
        }
    )

    fig.update_traces(marker=dict(size=14))

    fig.update_layout(
        plot_bgcolor='white',
        paper_bgcolor='white',

        # Axes in a full frame
        xaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True  # top axis line
        ),
        yaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True  # right axis line
        ),

        margin=dict(l=60, r=60, t=60, b=60),
        legend=dict(font=dict(size=14)),

        coloraxis_colorbar=dict(title="Di (Å)")
    )

    fig.show()
    fig.write_html(f"{csv_path.strip('.csv')}_Di.html")


def plot_1_cage_per_cell_PLD(csv_path):
    df = pd.read_csv(csv_path)

    fig = px.scatter(
        df,
        x="density",
        y="energy",
        color="PLD",
        color_continuous_scale="viridis",
        hover_data=["idx", "Di"],
        title=f'{csv_path.strip(".csv")}_PLD',
        labels={
            "density": "Density/ g cm\u207b\u00b3",
            "energy": "Energy per cage/ eV",
            "PLD": "PLD (Å)"
        }
    )

    fig.update_traces(marker=dict(size=14))

    fig.update_layout(
        plot_bgcolor='white',
        paper_bgcolor='white',

        xaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),
        yaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),

        legend=dict(font=dict(size=14)),
        margin=dict(l=60, r=60, t=60, b=60),
        coloraxis_colorbar=dict(title="PLD (Å)")
    )

    fig.show()
    fig.write_html(f"{csv_path.strip('.csv')}_PLD.html")


# Run your plots
for csv in [
    "OH_SO4_1_cage_janus_all.csv",
    "OH_PF6_1_cage_janus_all.csv",
    "SO3_gua_1_cage_janus_all.csv",
    "SO3_NMe4_1_cage_janus_all.csv"
]:
    plot_1_cage_per_cell_Di(csv)
    plot_1_cage_per_cell_PLD(csv)
```

```python
density_label = "Density/ g cm\u207b\u00b3"   # g cm⁻³
energy_label  = "Energy per cage/ eV"


def plot_2_cages_per_cell_Di(csv_path):
    df = pd.read_csv(csv_path)
    df['energy_scaled'] = df['energy'] / 2  # Scale energy per cage

    fig = px.scatter(
        df,
        x="density",
        y="energy_scaled",
        color="Di",
        color_continuous_scale="viridis",
        hover_data=["idx", "spg", "PLD"],
        title=f'{csv_path.strip(".csv")}_Di',
    )

    fig.update_traces(marker=dict(size=14))

    fig.update_layout(
        # Axis titles (unicode)
        xaxis_title=density_label,
        yaxis_title=energy_label,
        coloraxis_colorbar=dict(title="Di (Å)"),

        # Remove all background
        plot_bgcolor='white',
        paper_bgcolor='white',

        # Axes frame
        xaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),
        yaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),

        legend=dict(font=dict(size=14)),
        margin=dict(l=60, r=60, t=60, b=60),
    )

    fig.show()
    fig.write_html(f"{csv_path.strip('.csv')}_Di.html")


def plot_2_cages_per_cell_PLD(csv_path):
    df = pd.read_csv(csv_path)
    df['energy_scaled'] = df['energy'] / 2  # Scale energy per cage

    fig = px.scatter(
        df,
        x="density",
        y="energy_scaled",
        color="PLD",
        color_continuous_scale="viridis",
        hover_data=["idx", "spg", "Di"],
        title=f'{csv_path.strip(".csv")}_PLD',
    )

    fig.update_traces(marker=dict(size=14))

    fig.update_layout(
        # Axis titles (unicode)
        xaxis_title=density_label,
        yaxis_title=energy_label,
        coloraxis_colorbar=dict(title="PLD (Å)"),

        # Remove background
        plot_bgcolor='white',
        paper_bgcolor='white',

        # Axes as frame
        xaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),
        yaxis=dict(
            showline=True,
            linecolor='black',
            showgrid=False,
            ticks='outside',
            tickfont=dict(size=16),
            mirror=True
        ),

        legend=dict(font=dict(size=14)),
        margin=dict(l=60, r=60, t=60, b=60),
    )

    fig.show()
    fig.write_html(f"{csv_path.strip('.csv')}_PLD.html")
```
