```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_energy_vs_density(
    spg_list=[1, 2, 3, 9, 14, 19, 33],
    file_pattern='so3_na_{i}.csv',
    energy_col='Energies',
    density_col='Densities',
    energy_scale=96.5,
    energy_divisor=4,
    energy_min=9000,
    energy_max=10000,
    density_max=1.5,
    space_groups=None,
    output_csv='combined.csv',
    output_png='combined.png',
    title='Energy vs density diagram for the Cage 2 crystals',
    xlabel="\n Density / g cm\u207B\u00B3",
    ylabel="Relative total energy / (kJ mol\u207B\u00B9 of cage)",
    figsize=(10, 6),
):
    if space_groups is None:
        space_groups = [
            'triclinic P1', 'triclinic P\u00AF1',
            'monoclinic P2', 'monoclinic Cc',
            'monoclinic P2\u2081/c', 'orthorhombic P2\u2081 2\u2081 2\u2081',
            'orthorhombic Pna2\u2081'
        ]

    dfs = []
    for i in spg_list:
        filepath = file_pattern.format(i=i)
        df = pd.read_csv(filepath).iloc[:, 1:]
        df[energy_col] = (df[energy_col] * energy_scale) / energy_divisor
        df = df[
            (df[energy_col] < energy_max) &
            (df[energy_col] > energy_min) &
            (df[density_col] < density_max)
        ]
        dfs.append(df)

    combined_df = pd.concat(dfs, axis=1)
    combined_df.to_csv(output_csv, index=False)
    print(combined_df)

    fig, ax = plt.subplots(figsize=figsize)
    for idx, i in enumerate(range(0, combined_df.shape[1], 2)):
        label = space_groups[idx] if idx < len(space_groups) else f'Set {idx+1}'
        ax.scatter(combined_df.iloc[:, i + 1], combined_df.iloc[:, i], label=label)

    ax.set_xlabel(xlabel, size=14)
    ax.set_ylabel(ylabel, size=14)
    ax.legend(title='Space group')
    ax.grid(False)
    plt.suptitle(title, y=-0.05)
    plt.tight_layout()
    plt.savefig(output_png, bbox_inches='tight')
    plt.show()

    return combined_df
```
