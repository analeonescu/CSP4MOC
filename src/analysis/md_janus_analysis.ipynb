```python
from pathlib import Path
import pandas as pd

folder = Path("./OH_SO4_deprotonated_md")
energy_dic = {}

# Step 1: parse all files
for file in folder.glob("*.dat"):
    filename = file.name
    key = filename  # use full filename as column name

    file_dic = {}
    with file.open("r") as f:
        for line in f.readlines()[1:]:  # skip header
            parts = line.split()
            if len(parts) < 7:
                continue
            step = int(parts[0])
            energy = float(parts[6])
            file_dic[step] = energy

    if file_dic:
        energy_dic[key] = file_dic

# Step 2: convert to DataFrame
df = pd.DataFrame(energy_dic).sort_index()
df.index.name = "Step"

# Step 3: save to CSV
df.to_csv("OH_SO4_deprotonated_md.csv")
print("CSV saved successfully!")

```

```python
import plotly.graph_objects as go
def plot_md(file):
    sys_name = file.replace('_md.csv', '')
    df = pd.read_csv(file)
    fig = go.Figure()

    for col in df.columns[1:]:
        short_name = col.replace(f'xtal_{sys_name}_', '').rsplit('-nvt', 1)[0]
        fig.add_trace(go.Scatter(
            x=df['Step'],
            y=df[col],
            mode='lines+markers',
            name=short_name
        ))

    fig.update_layout(
        title=f'Energy vs Step for {file}',
        xaxis_title='Step no.',
        yaxis_title='Energy per atom/ eV',
        template='plotly_white',
        width=900,
        height=500
        )
    fig.show()
    fig.write_html(file.replace('.csv', '.html'))


plot_md('SO3_gua_md.csv')
plot_md('OH_PF6_md.csv')
plot_md('OH_SO4_deprotonated_md.csv')
plot_md('SO3_NMe4_md.csv')

```
