```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import numpy as np
```

```python
def plot_energy_density(
    csv_df1,
    csv_df2,
    energy_offset = 0,
    num_cages = 1
):
    """
    Plot Energy vs Density for two CSV files, coloured by Di and PLD.

    Args:
        csv_df1 (str): Path to the first CSV file (geomopt).
        csv_df2 (str): Path to the second CSV file (md+geomopt).
        energy_offset (float): Energy offset to apply before conversion (i.e. max physical energy).
        num_cages (int): Number of cages for energy normalisation (1, 2 or 4).
    """

    df1 = pd.read_csv(csv_df1)
    df2 = pd.read_csv(csv_df2)

    cols = ["idx", "energy", "density", "Di", "PLD"]
    df1[cols] = df1[cols].apply(pd.to_numeric, errors="coerce")
    df2[cols] = df2[cols].apply(pd.to_numeric, errors="coerce")

    df1["energy"] = (df1["energy"]/num_cages - energy_offset)*96.5
    df2["energy"] = (df2["energy"]/num_cages - energy_offset)*96.5

    # initialise default settings for all plots
    common_layout = dict(
        plot_bgcolor="white",
        paper_bgcolor="white",
        showlegend=False,
    )

    common_axis = dict(
        showgrid=False,
        zeroline=False,
        showline=True,
        mirror=True,
        ticks="outside",
        tickwidth=1,
        ticklen=6,
        linecolor="black"
    )

    di_min = min(df1["Di"].min(), df2["Di"].min())
    di_max = max(df1["Di"].max(), df2["Di"].max())

    fig_di = go.Figure()

    fig_di.add_trace(
        go.Scatter(
            x=df1["density"],
            y=df1["energy"],
            mode="markers",
            name="geomopt",
            marker=dict(
                symbol="circle",
                size=8,
                color=df1["Di"],
                colorscale="Viridis",
                cmin=di_min,
                cmax=di_max,
                colorbar=dict(title="Di"),
                opacity=0.8
            ),
            customdata=df1[["idx", "Di"]].values,
            hovertemplate=(
                "idx: %{customdata[0]}<br>"
                "Energy: %{y}<br>"
                "Density: %{x}<br>"
                "Di: %{customdata[1]}"
            )
        )
    )

    fig_di.add_trace(
        go.Scatter(
            x=df2["density"],
            y=df2["energy"],
            mode="markers",
            name="md+geomopt",
            marker=dict(
                symbol="star",
                size=12,
                color=df2["Di"],
                colorscale="Viridis",
                cmin=di_min,
                cmax=di_max,
                showscale=False,
                opacity=0.8,
                line=dict(
                    color="red",
                    width=2.5
                )
            ),
            customdata=df2[["idx", "Di"]].values,
            hovertemplate=(
                "idx: %{customdata[0]}<br>"
                "Energy: %{y}<br>"
                "Density: %{x}<br>"
                "Di: %{customdata[1]}"
            )
        )
    )

    fig_di.update_layout(
        title="Energy vs Density (colored by Di)",
        xaxis_title="Density/ g cm⁻³",
        yaxis_title="Cohesive energy/ kJ mol⁻¹",
        **common_layout
    )

    fig_di.update_xaxes(**common_axis)
    fig_di.update_yaxes(**common_axis)

    fig_di.show()

    pld_min = min(df1["PLD"].min(), df2["PLD"].min())
    pld_max = max(df1["PLD"].max(), df2["PLD"].max())

    fig_pld = go.Figure()

    fig_pld.add_trace(
        go.Scatter(
            x=df1["density"],
            y=df1["energy"],
            mode="markers",
            name="geomopt",
            marker=dict(
                symbol="circle",
                size=8,
                color=df1["PLD"],
                colorscale="Viridis",
                cmin=pld_min,
                cmax=pld_max,
                colorbar=dict(title="PLD"),
                opacity=0.8
            ),
            customdata=df1[["idx", "PLD"]].values,
            hovertemplate=(
                "idx: %{customdata[0]}<br>"
                "Energy: %{y}<br>"
                "Density: %{x}<br>"
                "PLD: %{customdata[1]}"
            )
        )
    )

    fig_pld.add_trace(
        go.Scatter(
            x=df2["density"],
            y=df2["energy"],
            mode="markers",
            name="md+geomopt",
            marker=dict(
                symbol="star",
                size=12,
                color=df2["PLD"],
                colorscale="Viridis",
                cmin=pld_min,
                cmax=pld_max,
                showscale=False,
                opacity=0.8,
                line=dict(
                    color="red",
                    width=2.5
                )
            ),
            customdata=df2[["idx", "PLD"]].values,
            hovertemplate=(
                "idx: %{customdata[0]}<br>"
                "Energy: %{y}<br>"
                "Density: %{x}<br>"
                "PLD: %{customdata[1]}"
            )
        )
    )

    fig_pld.update_layout(
        title="Energy vs Density (colored by PLD)",
        xaxis_title="Density/ g cm⁻³",
        yaxis_title="Cohesive energy/ kJ mol⁻¹",
        **common_layout
    )

    fig_pld.update_xaxes(**common_axis)
    fig_pld.update_yaxes(**common_axis)

    fig_pld.show()
```

```python
def plot_energy_density_common_idx_colored(
    csv_df1,
    csv_df2,
    energy_offset=0,
    num_cages=1
):

    df1 = pd.read_csv(csv_df1)
    df2 = pd.read_csv(csv_df2)

    cols = ["idx", "energy", "density"]
    df1[cols] = df1[cols].apply(pd.to_numeric, errors="coerce")
    df2[cols] = df2[cols].apply(pd.to_numeric, errors="coerce")

    # Energy normalisation
    df1["energy"] = (df1["energy"]/num_cages - energy_offset)*96.5
    df2["energy"] = (df2["energy"]/num_cages - energy_offset)*96.5

    # Keep common idx only
    common_idx = sorted(set(df1["idx"]).intersection(df2["idx"]))

    # Manual color palette (extend if needed)
    colors = [
        "#1f77b4", "#ff7f0e", "#2ca02c", "#d62728",
        "#9467bd", "#8c564b", "#e377c2", "#7f7f7f",
        "#bcbd22", "#17becf"
    ]

    fig = go.Figure()

    for i, idx in enumerate(common_idx):

        c = colors[i % len(colors)]

        p1 = df1[df1["idx"] == idx]
        p2 = df2[df2["idx"] == idx]

        # geomopt point (circle)
        fig.add_trace(
            go.Scatter(
                x=p1["density"],
                y=p1["energy"],
                mode="markers",
                marker=dict(
                    symbol="circle",
                    size=12,
                    color=c
                ),
                name=f"idx {idx}",
                showlegend=False,
                hovertemplate=(
                    f"idx: {idx}<br>"
                    "Energy: %{y}<br>"
                    "Density: %{x}<br>"
                    "geomopt"
                )
            )
        )

        # md+geomopt point (star)
        fig.add_trace(
            go.Scatter(
                x=p2["density"],
                y=p2["energy"],
                mode="markers",
                marker=dict(
                    symbol="star",
                    size=18,  # bigger
                    color=c,
                    line=dict(color=c, width=2.5)  # contour same as fill
                ),
                name=f"idx {idx}",
                showlegend=False,
                hovertemplate=(
                    f"idx: {idx}<br>"
                    "Energy: %{y}<br>"
                    "Density: %{x}<br>"
                    "md+geomopt"
                )
            )
        )

    common_layout = dict(
        plot_bgcolor="white",
        paper_bgcolor="white",
    )

    common_axis = dict(
        showgrid=False,
        zeroline=False,
        showline=True,
        mirror=True,
        ticks="outside",
        tickwidth=1,
        ticklen=6,
        linecolor="black"
    )

    fig.update_layout(
        title="Energy vs Density (common idx, colored by pair)",
        xaxis_title="Density / g cm⁻³",
        yaxis_title="Cohesive energy / kJ mol⁻¹",
        **common_layout
    )

    fig.update_xaxes(**common_axis)
    fig.update_yaxes(**common_axis)

    fig.show()
```

```python
def plot_energy_density_multi_idx(
    datasets,
    energy_offset=0,
    highlight_symbol="x"   # symbol for 1-cage matching idx
):
    """
    Big CSVs:
        - one fixed color per dataset
    Top CSVs:
        - colour = idx (global)
        - symbol = dataset
    Extra:
        - matching idx from 1-cage big CSV
          plotted with a special symbol
    """

    dfs_all = []
    dfs_top = []

    for d in datasets:
        df_all = pd.read_csv(d["all"])
        df_top = pd.read_csv(d["top"])

        cols = ["idx", "energy", "density", "Di", "PLD"]
        df_all[cols] = df_all[cols].apply(pd.to_numeric, errors="coerce")
        df_top[cols] = df_top[cols].apply(pd.to_numeric, errors="coerce")

        df_all["energy"] = (df_all["energy"]/d["num_cages"] - energy_offset)*96.5
        df_top["energy"] = (df_top["energy"]/d["num_cages"] - energy_offset)*96.5

        df_top["symbol"] = d["top_symbol"]
        df_top["label"] = d["label"]
        df_all["label"] = d["label"]
        df_all["set_color"] = d["set_color"]

        dfs_all.append(df_all)
        dfs_top.append(df_top)

    all_idx = pd.concat(dfs_top)["idx"].unique()
    all_idx = np.sort(all_idx)

    cmap = px.colors.qualitative.Dark24
    idx_color = {
        idx: cmap[i % len(cmap)]
        for i, idx in enumerate(all_idx)
    }

    fig = go.Figure()

    for df in dfs_all:
        fig.add_trace(
            go.Scatter(
                x=df["density"],
                y=df["energy"],
                mode="markers",
                name=f"{df['label'].iloc[0]} all",
                marker=dict(
                    size=7,
                    color=df["set_color"].iloc[0],
                    opacity=0.5
                ),
                showlegend=True
            )
        )

    for df in dfs_top:
        fig.add_trace(
            go.Scatter(
                x=df["density"],
                y=df["energy"],
                mode="markers",
                name=f"{df['label'].iloc[0]} top",
                marker=dict(
                    symbol=df["symbol"].iloc[0],
                    size=14,
                    color=[idx_color[i] for i in df["idx"]],
                    line=dict(color="black", width=1)
                ),
                customdata=df[["idx", "Di", "PLD"]].values,
                hovertemplate=(
                    "idx: %{customdata[0]}<br>"
                    "Energy: %{y}<br>"
                    "Density: %{x}<br>"
                    "Di: %{customdata[1]}<br>"
                    "PLD: %{customdata[2]}"
                )
            )
        )

    df_1c_all = next(
        d for d in dfs_all if d["label"].iloc[0] == "1 cage"
    )
    df_1c_top = next(
        d for d in dfs_top if d["label"].iloc[0] == "1 cage"
    )

    match_idx = df_1c_top["idx"].unique()
    df_match = df_1c_all[df_1c_all["idx"].isin(match_idx)]

    fig.add_trace(
        go.Scatter(
            x=df_match["density"],
            y=df_match["energy"],
            mode="markers",
            name="1 cage matching idx",
            marker=dict(
                symbol=highlight_symbol,
                size=16,
                color=[idx_color[i] for i in df_match["idx"]],
                line=dict(color="black", width=2)
            ),
            customdata=df_match[["idx"]].values,
            hovertemplate=(
                "MATCH idx: %{customdata[0]}<br>"
                "Energy: %{y}<br>"
                "Density: %{x}"
            )
        )
    )

    fig.update_layout(
        title="Energy vs Density",
        xaxis_title="Density / g cm⁻³",
        yaxis_title="Cohesive energy / kJ mol⁻¹",
        plot_bgcolor="white",
        paper_bgcolor="white"
    )

    axis_style = dict(
        showgrid=False,
        zeroline=False,
        showline=True,
        mirror=True,
        ticks="outside",
        tickwidth=1,
        ticklen=6,
        linecolor="black"
    )

    fig.update_xaxes(**axis_style)
    fig.update_yaxes(**axis_style)

    fig.show()

```

```python
datasets = [
    {
        "all": "./OH_PF6_md/OH_PF6_4_cages_janus_all.csv",
        "top": "./OH_PF6_md/OH_PF6_4cc_top.csv",
        "num_cages": 4,
        "top_symbol": "star",
        "label": "4 cages",
        "set_color": "#1f77b4"   # blue
    },
    {
        "all": "./OH_PF6_md/OH_PF6_2_cages_janus_all.csv",
        "top": "./OH_PF6_md/OH_PF6_2cc_top.csv",
        "num_cages": 2,
        "top_symbol": "diamond",
        "label": "2 cages",
        "set_color": "#2ca02c"   # green
    },
    {
        "all": "./OH_PF6_md/OH_PF6_1_cage_janus_all.csv",
        "top": "./OH_PF6_md/OH_PF6_1cc_top.csv",
        "num_cages": 1,
        "top_symbol": "cross",
        "label": "1 cage",
        "set_color": "#ff7f0e"   # orange
    }
]

plot_energy_density_multi_idx(
    datasets,
    energy_offset=-2288.04
)
```
