```python
import pandas as pd

df = pd.read_csv("titanic.csv", index_col="PassengerId")

# Quick inspection tools
df.head(10)      # First 10 rows
df.info()        # Overview of data types and missing values
df.describe()    # Statistics for numerical columns
df.shape         # Get (rows, columns)
```

```python
age_series = df["Age"]             # Returns a Series
subset_df = df[["Age", "Sex"]]     # Returns a DataFrame

# Selecting by position (.iloc)
df.iloc[0:5, :3]                   # First 5 rows and first 3 columns

# Selecting by label (.loc)
# Useful for filtering specific rows and columns by name
df.loc[1:10, ["Age", "Fare"]]
```

```python
# Single condition: passengers older than 30
df[df["Age"] > 30]

# Multiple conditions: Use & (and), | (or)
# Important: Always wrap conditions in parentheses
mask = (df["Sex"] == "female") & (df["Age"] < 18)
df[mask]

# Filtering with methods
df[df["Embarked"].isin(["C", "S"])]
df[df["Cabin"].notna()]  # Rows where Cabin is not empty
```

```python
# Creating new columns via broadcasting
df["Fare_Adjusted"] = df["Fare"] * 1.1

# Replacing specific values
df["Sex"].replace({"male": 0, "female": 1}, inplace=True)

# The Best Practice: Using .loc for assignment
# This avoids the 'SettingWithCopyWarning'
df.loc[df["Age"] < 1, "Age"] = 1

# Renaming columns
df.rename(columns={"Pclass": "TicketClass"}, inplace=True)
```

```python
df.index                # Access row labels
df.columns              # Access column labels
df.index.name = "ID"    # Assign a name to the index level

# Moving the index back to a column
df.reset_index(inplace=True)
```

```python
# Sorting by values
df.sort_values(by=["Pclass", "Age"], ascending=[True, False])

# Summary counts
df["Embarked"].value_counts()    # Frequency of each category
df["Age"].nunique()              # Number of unique ages
```

```python
# Use the .str accessor to apply string methods to a whole column
df[df["Name"].str.contains("Master")]
df["Name"].str.upper()
df["Ticket"].str.strip()
```
